# FashionMNIST – Khám phá dữ liệu

Notebook này dùng để kiểm tra bộ dữ liệu FashionMNIST, xem kích thước dữ liệu và hiển thị một số ảnh quần áo.

> Chạy các ô theo thứ tự từ trên xuống. Lần đầu, `torchvision` có thể tải dữ liệu về máy.

```python
import torch
import matplotlib.pyplot as plt
from torchvision import datasets, transforms

transform = transforms.ToTensor()
train_dataset = datasets.FashionMNIST(
    root="./data", train=True, download=True, transform=transform
)
test_dataset = datasets.FashionMNIST(
    root="./data", train=False, download=True, transform=transform
)

classes = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]

print("PyTorch version:", torch.__version__)
print("So anh huan luyen:", len(train_dataset))
print("So anh kiem tra:", len(test_dataset))
print("So loai quan ao:", len(classes))
```

```python
# Xem mot anh mau va nhan tuong ung
image, label = train_dataset[0]
print("Kich thuoc anh:", tuple(image.shape))
print("Nhan:", classes[label])

plt.figure(figsize=(3, 3))
plt.imshow(image.squeeze(), cmap="gray")
plt.title(classes[label])
plt.axis("off")
plt.show()
```

```python
# Hien thi 10 anh dau tien
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for i, ax in enumerate(axes.flat):
    image, label = train_dataset[i]
    ax.imshow(image.squeeze(), cmap="gray")
    ax.set_title(classes[label])
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Nhận xét

- FashionMNIST có 60.000 ảnh huấn luyện và 10.000 ảnh kiểm tra.
- Mỗi ảnh là ảnh xám kích thước 28 × 28 pixel.
- Bộ dữ liệu có 10 nhãn quần áo và phụ kiện.
- Notebook này chỉ khám phá dữ liệu; phần huấn luyện mô hình nằm trong `main.py`.
